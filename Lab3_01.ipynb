```python
import numpy as np

def gauss_jordan(A, b, tol=1e-12):
    """
    Solves Ax = b with the Gauss-Jordan method (with partial pivoting).
    Returns the solution vector, or None if there is no unique solution.
    """
    A = np.array(A, dtype=float)
    b = np.array(b, dtype=float).reshape(-1)
    n = A.shape[0]
    if A.shape != (n, n) or b.size != n:
        raise ValueError("A must be n x n and b must have n elements.")

    M = np.hstack([A, b.reshape(-1, 1)])
    scale = max(1.0, np.max(np.abs(A)))

    for k in range(n):

        p = k + np.argmax(np.abs(M[k:, k]))
        if abs(M[p, k]) < tol * scale:
            return None
        if p != k:
            M[[k, p]] = M[[p, k]]

        M[k] = M[k] / M[k, k]

        for i in range(n):
            if i != k:
                M[i] = M[i] - M[i, k] * M[k]

    return M[:, -1]
```

```python
def test_system(n, seed=0):
    rng = np.random.default_rng(seed)
    A = rng.uniform(-10, 10, (n, n))
    b = rng.uniform(-10, 10, n)

    x = gauss_jordan(A, b)
    x_np = np.linalg.solve(A, b)

    print(f"System with n = {n}")
    print("Gauss-Jordan    :", np.round(x, 6))
    print("np.linalg.solve :", np.round(x_np, 6))
    print("Max difference  :", np.max(np.abs(x - x_np)))
    print("Residual ||Ax-b||:", np.linalg.norm(A @ x - b))
    print("Match:", np.allclose(x, x_np), "\n")
```

```python
## n=5
test_system(5)
```

```python
## n=10
test_system(10, seed=1)
```

```python
# Row 3 = row 1 + row 2,
## ==== > so the matrix is singular
A_sing = [[1, 2, 3],
          [4, 5, 6],
          [5, 7, 9]]
b_sing = [1, 2, 3]

res = gauss_jordan(A_sing, b_sing)
if res is None:
    print("The system does not have a unique solution.")
else:
    print("Solution:", res)
```
